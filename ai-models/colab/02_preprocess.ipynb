```python
from pathlib import Path
import os
import subprocess
import sys

REPO_URL = "https://github.com/YoungHyyy/18_10123154_12523066_CDUTV.git"
repo_root = next(
    (
        candidate
        for candidate in (Path.cwd(), *Path.cwd().parents)
        if (candidate / "ai-models" / "data" / "dataset.zip").is_file()
    ),
    None,
)
if repo_root is None:
    repo_root = Path("/content/18_10123154_12523066_CDUTV")
    if not (repo_root / "ai-models" / "data" / "dataset.zip").is_file():
        repo_root.parent.mkdir(parents=True, exist_ok=True)
        subprocess.run(["git", "clone", REPO_URL, str(repo_root)], check=True)

os.chdir(repo_root)
sys.path.insert(0, str(repo_root / "ai-models" / "src"))

import sklearn, numpy, pandas
print("sklearn", sklearn.__version__, "| numpy", numpy.__version__, "| pandas", pandas.__version__)
print("Repo:", repo_root)
```

```python
from preprocess import load_data, split, build_schema

X, y = load_data()
print("X:", X.shape, "| trùng lặp đã bỏ, thiếu:", X.isna().sum().sum())
print("Tỉ lệ ác tính toàn bộ:", round(y.mean(), 3))

X_train, X_test, y_train, y_test = split(X, y)
print("Train:", X_train.shape, "| Test:", X_test.shape)
print("Tỉ lệ ác tính - train:", round(y_train.mean(), 3), "| test:", round(y_test.mean(), 3))
```
